# Evaluation

```python
import pickle
from pathlib import Path

import eval_tools

def parse_test_scenarios(split_file="noa_nn/train_test_split.txt"):
    scenarios = []
    in_test = False
    with open(split_file, "r", encoding="utf-8") as f:
        for raw in f:
            line = raw.strip()
            if not line or line.startswith("#"):
                continue
            if line.startswith("[") and line.endswith("]"):
                in_test = (line == "[TEST_SCENARIOS]")
                continue
            if in_test:
                scenarios.append(line)
    return scenarios

exp_dir = Path("experiments/NOA_NN")
scenarios_dir = Path("scenarios")
out_dir = Path("eval/NOA_NN")
out_dir.mkdir(parents=True, exist_ok=True)

scenario_ids = parse_test_scenarios("noa_nn/train_test_split_model1.txt")

errs_dict = {}
skipped = []

for sid in scenario_ids:
    hypfile = exp_dir / sid / "hyp.npy"
    q_annot = scenarios_dir / sid / "query.beats"
    r_annot = scenarios_dir / sid / "ref.beats"

    if not hypfile.exists():
        skipped.append((sid, "missing_hyp"))
        continue

    err = eval_tools.eval_alignment_single(str(hypfile), str(q_annot), str(r_annot))

    # Guard against malformed returns
    if err is None or (isinstance(err, tuple) and any(x is None for x in err)):
        skipped.append((sid, "invalid_err"))
        continue

    err = np.asarray(err, dtype=float).reshape(-1)
    err = err[~np.isnan(err)]
    if err.size == 0:
        skipped.append((sid, "empty_err"))
        continue

    errs_dict[sid] = err

with open(out_dir / "errs.pkl", "wb") as f:
    pickle.dump(errs_dict, f)

print(f"Saved: {out_dir / 'errs.pkl'}")
print(f"Evaluated scenarios: {len(errs_dict)}")
print(f"Skipped scenarios: {len(skipped)}")
if skipped:
    print("Skipped detail:", skipped[:10])
```
